Write a Python program using Pandas to create a DataFrame containing student names, ages, and marks, and display the DataFrame and students who scored more than 75 marks.


```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Kiran", "Sneha"],
    "Age": [21, 22, 20, 23],
    "Marks": [85, 68, 92, 78]
}

df = pd.DataFrame(data)

print("Student DataFrame:")
print(df)

print("\nStudents who scored more than 75 marks:")
print(df[df["Marks"] > 75])
```

13. Find Factorial of a Number

PROMPT:

Write a Python program to calculate the factorial of a given positive integer using a loop.

```python
num = 5
factorial = 1

for i in range(1, num + 1):
    factorial *= i

print("Factorial:", factorial)
```

14. Find Duplicate Elements in a List

PROMPT:

Write a Python program to find and display duplicate elements in a given list.

```python
numbers = [10, 20, 30, 20, 40, 10, 50]

duplicates = []

for num in numbers:
    if numbers.count(num) > 1 and num not in duplicates:
        duplicates.append(num)

print("Duplicate elements:", duplicates)
```

15. Count Character Frequency

PROMPT:

Write a Python program to count the frequency of each character in a given string using a dictionary.

```python
text = "python"
frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

print("Character Frequency:", frequency)
```
