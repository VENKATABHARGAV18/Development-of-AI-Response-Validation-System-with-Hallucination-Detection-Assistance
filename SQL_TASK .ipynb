```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

print("SQLite database connected successfully.")
```

1: CREATE TABLE AND INSERT DATA

PROMPT:

Write SQL queries to create a students table with student ID, name, age, and marks columns, and insert three student records into the table.

```python
cursor.execute("""
CREATE TABLE students (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER
)
""")

cursor.executemany("""
INSERT INTO students (student_id, name, age, marks)
VALUES (?, ?, ?, ?)
""", [
    (1, "Ravi", 21, 85),
    (2, "Priya", 22, 90),
    (3, "Kiran", 20, 75)
])

conn.commit()

query = "SELECT * FROM students"
df = pd.read_sql_query(query, conn)
print(df.to_string(index=False))
```

5: AGGREGATE FUNCTIONS

PROMPT:

Write an SQL query to calculate the total number of students, average marks, highest marks, and lowest marks from the students table.

```python
query = """
SELECT
    COUNT(*) AS total_students,
    AVG(marks) AS average_marks,
    MAX(marks) AS highest_marks,
    MIN(marks) AS lowest_marks
FROM students
"""

df = pd.read_sql_query(query, conn)
print(df.to_string(index=False))
```

6: GROUP BY AND HAVING

```python
query = """
SELECT age, COUNT(*) AS total_students
FROM students
GROUP BY age
ORDER BY age
"""

df = pd.read_sql_query(query, conn)
print(df.to_string(index=False))
```

7: JOINS

PROMPT:

Write SQL queries to create a courses table, insert three course records using the existing student IDs, and display student names along with their course names using INNER JOIN, LEFT JOIN, and RIGHT JOIN.

```python
# Create courses table
cursor.execute("""
CREATE TABLE courses (
    student_id INTEGER,
    course_name TEXT
)
""")

cursor.executemany("""
INSERT INTO courses (student_id, course_name)
VALUES (?, ?)
""", [
    (1, "Python"),
    (2, "SQL"),
    (3, "Data Science")
])

conn.commit()

# INNER JOIN
query = """
SELECT students.student_id, students.name, courses.course_name
FROM students
INNER JOIN courses
ON students.student_id = courses.student_id
"""

df = pd.read_sql_query(query, conn)
print("INNER JOIN:")
print(df.to_string(index=False))

# LEFT JOIN
query = """
SELECT students.student_id, students.name, courses.course_name
FROM students
LEFT JOIN courses
ON students.student_id = courses.student_id
"""

df = pd.read_sql_query(query, conn)
print("\nLEFT JOIN:")
print(df.to_string(index=False))

# RIGHT JOIN
query = """
SELECT students.student_id, students.name, courses.course_name
FROM students
RIGHT JOIN courses
ON students.student_id = courses.student_id
"""

df = pd.read_sql_query(query, conn)
print("\nRIGHT JOIN:")
print(df.to_string(index=False))
```

8: SUBQUERIES

PROMPT:

Write SQL queries to find students whose marks are greater than the average marks and identify the student with the highest marks using subqueries.

```python
# Query 1: Students scoring above average
query1 = """
SELECT student_id, name, marks
FROM students
WHERE marks > (
    SELECT AVG(marks)
    FROM students
)
"""

df1 = pd.read_sql_query(query1, conn)
print("Students Scoring Above Average:")
print(df1.to_string(index=False))

# Query 2: Student with the highest marks
query2 = """
SELECT student_id, name, marks
FROM students
WHERE marks = (
    SELECT MAX(marks)
    FROM students
)
"""

df2 = pd.read_sql_query(query2, conn)
print("\nStudent with Highest Marks:")
print(df2.to_string(index=False))
```

9: UPDATE, DELETE AND CONSTRAINTS

PROMPT:

Write SQL queries to update a student's marks, delete a student record from a test table, and create a table using PRIMARY KEY, NOT NULL, and UNIQUE constraints.

```python
# Create a test table using existing student records
cursor.execute("""
CREATE TABLE students_test AS
SELECT * FROM students
""")

# Query 1: Update marks in the test table
cursor.execute("""
UPDATE students_test
SET marks = 88
WHERE student_id = 1
""")

df1 = pd.read_sql_query(
    "SELECT * FROM students_test WHERE student_id = 1", conn
)
print("After UPDATE:")
print(df1.to_string(index=False))

# Query 2: Delete a record from the test table
cursor.execute("""
DELETE FROM students_test
WHERE student_id = 3
""")

df2 = pd.read_sql_query("SELECT * FROM students_test", conn)
print("\nAfter DELETE:")
print(df2.to_string(index=False))

# Query 3: Create a table with constraints
cursor.execute("""
CREATE TABLE employees (
    employee_id INTEGER PRIMARY KEY,
    employee_name TEXT NOT NULL,
    email TEXT UNIQUE
)
""")

cursor.executemany("""
INSERT INTO employees (employee_id, employee_name, email)
VALUES (?, ?, ?)
""", [
    (101, "Ravi", "ravi@example.com"),
    (102, "Priya", "priya@example.com")
])

df3 = pd.read_sql_query("SELECT * FROM employees", conn)
print("\nEmployees Table:")
print(df3.to_string(index=False))

conn.commit()
```

10:Find the Second-Highest Marks

Write an SQL query to find the second-highest distinct marks from the existing students table.

```python
query = """
SELECT MAX(marks) AS second_highest_marks
FROM students
WHERE marks < (
    SELECT MAX(marks)
    FROM students
)
"""

df = pd.read_sql_query(query, conn)
print(df.to_string(index=False))
```

11:Find Duplicate Marks

PROMPT:

Write an SQL query to identify marks that appear more than once in the existing students table.

```python
query = """
SELECT marks, COUNT(*) AS occurrence_count
FROM students
GROUP BY marks
HAVING COUNT(*) > 1
"""

df = pd.read_sql_query(query, conn)
print(df.to_string(index=False))
```
